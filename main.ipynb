```python
from dataclasses import dataclass

print("INTELLIGENT AGENT BASED COURSE RECOMMENDATION SYSTEM\n")

# ------------------ PEAS MODEL ------------------
print("PEAS MODEL")
print("Performance Measure : Recommend best course")
print("Environment         : Online Course Platform")
print("Actuators           : Display Recommended Course")
print("Sensors             : Student Preferences\n")


# ------------------ STATE ------------------
@dataclass
class StudentState:
    interest: str
    difficulty: int
    budget: int


# ------------------ KNOWLEDGE BASE ------------------
courses = [
    {
        "name": "Artificial Intelligence",
        "interest": "AI",
        "difficulty": 8,
        "cost": 5000,
        "placement": 9,
        "credits": 4
    },
    {
        "name": "Data Science",
        "interest": "Data",
        "difficulty": 7,
        "cost": 4500,
        "placement": 8,
        "credits": 3
    },
    {
        "name": "Machine Learning",
        "interest": "AI",
        "difficulty": 9,
        "cost": 6000,
        "placement": 9,
        "credits": 5
    },
    {
        "name": "Web Development",
        "interest": "Web",
        "difficulty": 5,
        "cost": 3000,
        "placement": 7,
        "credits": 2
    }
]


# ================== CO1 ==================
print("\n========== CO1: AGENT BASED FILTER ==========\n")

interest = input("Enter Interest (AI/Data/Web): ")
difficulty = int(input("Preferred Difficulty (1-10): "))
budget = int(input("Enter Budget: "))

state = StudentState(interest, difficulty, budget)

recommended = []

for course in courses:
    print("Checking:", course["name"])

    if (
        course["interest"] == state.interest and
        course["difficulty"] <= state.difficulty and
        course["cost"] <= state.budget
    ):
        recommended.append(course)
        print("Result: Accepted\n")
    else:
        print("Result: Rejected\n")

print("CO1 Recommended Courses:")
if recommended:
    for c in recommended:
        print(c["name"])
else:
    print("No suitable course found")


# ================== CO2 ==================
print("\n========== CO2: SEARCH BASED ==========\n")

placement_importance = int(input("Placement importance (1-10): "))
interest_importance = int(input("Interest importance (1-10): "))
difficulty_preference = int(input("Preferred difficulty (1-10): "))

best_course = None
best_score = -999

for course in courses:
    score = (
        placement_importance * course["placement"] +
        interest_importance * (10 if course["interest"] == interest else 5) -
        abs(difficulty_preference - course["difficulty"])
    )

    print("Course:", course["name"], "| Score:", score)

    if score > best_score:
        best_score = score
        best_course = course

print("\nBest Course (CO2):", best_course["name"])


# ================== CO3 ==================
print("\n========== CO3: CONSTRAINT SATISFACTION ==========\n")

max_credits = int(input("Maximum credits allowed: "))

valid_courses = []

for course in courses:
    if course["credits"] <= max_credits:
        valid_courses.append(course)
        print(course["name"], "Accepted")
    else:
        print(course["name"], "Rejected (Too many credits)")

best_course = None
best_score = -999

for course in valid_courses:
    score = (
        placement_importance * course["placement"] +
        interest_importance * (10 if course["interest"] == interest else 5) -
        abs(difficulty_preference - course["difficulty"])
    )

    print("Course:", course["name"], "| Final Score:", score)

    if score > best_score:
        best_score = score
        best_course = course

print("\nFINAL RECOMMENDED COURSE:")
print("Course Name:", best_course["name"])
print("Score:", best_score)

print("\nReason:")

if best_course["placement"] >= 8:
    print("- High placement opportunities")

if best_course["interest"] == interest:
    print("- Matches your interest")

if best_course["credits"] <= max_credits:
    print("- Within credit limit")
```
